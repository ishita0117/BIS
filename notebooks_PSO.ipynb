```python
import numpy as np

f = lambda x: sum(x**2)
n, d = 20, 2
x = np.random.uniform(-10, 10, (n, d))
v = np.zeros((n, d))
p = x.copy()
g = p[np.argmin([f(i) for i in p])]
for _ in range(100):
    v = .5*v + 1.5*np.random.rand(n,d)*(p-x) + 1.5*np.random.rand(n,d)*(g-x)
    x += v
    for i in range(n):
        if f(x[i]) < f(p[i]): p[i] = x[i]
    g = p[np.argmin([f(i) for i in p])]
print("Best:", g, "Minimum:", f(g))
```
