```python
import numpy as np

f = lambda x: sum(x**2)
n, d = 20, 2
x = np.random.uniform(-10,10,(n,d))
for t in range(100):
    x = sorted(x, key=f)
    A, B, C = x[:3]
    a = 2-2*t/100
    r1, r2 = np.random.rand(3,d), np.random.rand(3,d)
    X = [A,B,C]
    y = sum(z-a*(2*r1[i]-1)*abs(2*r2[i]*z-x)
            for i,z in enumerate(X))/3
    x = np.clip(y,-10,10)
print("Best:", x, "Minimum:", f(x))
```
