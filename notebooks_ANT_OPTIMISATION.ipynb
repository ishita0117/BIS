```python
import numpy as np

d = np.array([[0,2,9,10],[2,0,6,4],[9,6,0,8],[10,4,8,0]])
n, a, b, rho = 10, 1, 2, .5
p = np.ones((4,4))
best, cost = None, float('inf')

for _ in range(100):
    for _ in range(n):
        path, city = [0], 0
        while len(path)<4:
            q = [i for i in range(4) if i not in path]
            w = [p[city][i]**a*(1/d[city][i])**b for i in q]
            city = np.random.choice(q, p=np.array(w)/sum(w))
            path.append(city)
        path.append(0)
        dist = sum(d[path[i]][path[i+1]] for i in range(4))
        if dist < cost: best, cost = path, dist
    p *= 1-rho
    for i in range(4):
        for j in range(4):
            p[best[i]][best[i+1]] += 1/cost
print("Best route:", best, "Distance:", cost)
```
