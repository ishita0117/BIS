```python
import numpy as np

f = lambda x: sum(x**2)
x = np.random.uniform(-10,10,(5,5,2))

for _ in range(100):
    y = x.copy()
    for i in range(5):
        for j in range(5):
            nb = [x[i,j]]
            for a,b in [(0,1),(0,-1),(1,0),(-1,0)]:
                if 0<=i+a<5 and 0<=j+b<5:
                    nb.append(x[i+a,j+b])
            best = min(nb,key=f)
            y[i,j] = best + np.random.randn(2)*.1
    x = y

best = min(x.reshape(-1,2),key=f)
print("Best:", best, "Minimum:", f(best))
```
