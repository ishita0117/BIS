```python
import numpy as np

f = lambda x: x*np.sin(10*np.pi*x)+1
p = np.random.rand(20)

for _ in range(100):
    p = sorted(p, key=f, reverse=True)
    q = p[:2]
    while len(q)<20:
        a,b = np.random.choice(p[:10],2)
        c = (a+b)/2
        if np.random.rand()<.2: c=np.random.rand()
        q.append(c)
    p=np.array(q)

print("Best:", p[0], "Maximum:", f(p[0]))
```
