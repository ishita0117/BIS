```python
import numpy as np

f = lambda x: sum(x**2)
n, d = 20, 2
x = np.random.uniform(-10,10,(n,d))
for _ in range(100):
    i = np.random.randint(n)
    y = x[i] + np.random.randn(d)*np.random.rand()
    if f(y)<f(x[i]): x[i]=y
    k = np.random.rand(n)<.25
    x[k]=np.random.uniform(-10,10,(sum(k),d))
print("Best:", x[np.argmin([f(i) for i in x])])
```
